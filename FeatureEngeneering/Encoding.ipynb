```python
import pandas as pd
from pathlib import Path

# Load Titanic dataset
data_path = next((p for p in [Path("../Data Processing/titanic.datatype"), Path("titanic.datatype"), Path.cwd() / "titanic.datatype"] if p.exists()), "../Data Processing/titanic.datatype")
titanic_df = pd.read_csv(data_path)

titanic_df.head()
```

Data Encoding

Data Encoding is a technique that is used to convert categorical columns into numerical ones so that they can be fitted by machine learning models which only take numerical data. It is an important pre-processing step in a machine-learning project.



```python
### Nominal Data  -> data without natural order
titanic_df.Sex
```

One- Hot Encoding

```python

# Pandas
pd.get_dummies(titanic_df, columns=["Sex","Embarked"], dtype=int).head(5)
```

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_encoder = OneHotEncoder(sparse_output=False)
```

```python
one_hot_encoder.fit_transform(
    titanic_df[["Sex","Embarked"]]
)
```

```python
one_hot_encoder.categories_
```

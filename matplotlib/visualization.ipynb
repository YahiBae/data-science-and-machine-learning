visualization of EDA

```python
import pandas as pd
from pathlib import Path

data_path = Path("pandas/titanic.datatype")
if not data_path.exists():
    data_path = Path("../pandas/titanic.datatype")

titanic_df = pd.read_csv(data_path)
```

```python
titanic_df.Survived.value_counts().plot(
    kind="bar",
    title="Survived Count",
    figsize=(10, 5),
    legend=True,
    xlabel="survived (0 = No, 1 = Yes)",
    ylabel="n0. of passengers",
    grid=False,
    yticks=list(range(0, 601, 50)),
    #colormap = "Set1"
    color = ["red", "green"]
)
```

```python
titanic_df.value_counts(subset = ["Survived", "Sex"]).plot(
    kind="bar",
        title="Survived Count",
        figsize=(10, 5),
        legend=True,
        xlabel="survived (0 = No, 1 = Yes)",
        ylabel="n0. of passengers",
        grid=False,
        yticks=list(range(0, 601, 50)),
        #colormap = "Set1"
        color = ["red", "green"]
)
```

```python
pd.crosstab(titanic_df["Survived"], titanic_df["Sex"]).plot(
    kind="bar",
    title="Survival by Sex",
    figsize=(10, 5),
    legend=True,
    xlabel="Survived (0 = No, 1 = Yes)",
    ylabel="Number of passengers",
    grid=False,
    color=["#d62728", "#2ca02c"]
)
```

2. pie chart

```python
group_by_embarked = titanic_df.groupby(by = "Embarked")
group_by_embarked.count()
```

```python
group_by_embarked.size().plot(
    kind="pie",
    title="Passenger Count by Embarkation Port",
    autopct="%1.1f%%",
    startangle=90
)
```

```python
crosstab = pd.crosstab(
  titanic_df.Survived,
  titanic_df.Pclass  
)
crosstab
```

```python
crosstab.T.plot(
    kind = "pie",
    subplots = True,
    figsize = (5, 5)
)
```

3. histogram

```python
titanic_df.Age.plot(
    kind = "hist",
    bins = 20,
    grid = True,
    figsize = (10, 5)
)

titanic_df.Age.mean()
```

```python
titanic_df.Age.plot(
    kind = "kde",
    ind = range(0, 81),
    grid = True,
    figsize = (10, 5)
)
```

```python
titanic_df[["Age", "Sex"]].plot(
    kind = "hist",
    by = "Sex",
    bins = 20,
    figsize = (10, 5),
)
```

```python
titanic_df[["Age", "Survived"]].plot(
    kind = "hist",
    by = "Survived",
    bins = 20,
    figsize = (10, 5),
)
```

```python
titanic_df[["Age", "Fare"]].plot(
    kind = "kde",
    figsize = (10, 5),
    ind = range(0, 200)
)
```

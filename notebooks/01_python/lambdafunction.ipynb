lambda expressions

```python
def power_up(base, power):
    return base ** power
```

```python
power_up(2, 3)
```

```python
power_up_lambda = lambda base, power: base ** power
```

```python
power_up_lambda(2, 3)
```

```python
area = lambda l, b, h: 2 * (l*b + b*h + l*h)

print(area(20, 22, 5))
```

property decorator in py


getter

```python
class Person:
    def __init__(self, name):
        self.full_name = name

    @property
    def name(self):
        print("Getting name...")
        return self.full_name


    
```

```python
obj = Person("John Doe")
```

```python
import uuid

class Person:
    def __init__(self, name):
        self.full_name = name

    @property
    def name(self):
        print("Getting name...")
        return self.full_name

    @property
    def id(self):
        print("Getting id...")
        return f"{self.full_name[-3:]}-{str(uuid.uuid4())[:8]}"
```

```python
obj = Person("Sagar Pun")
```

```python

obj.id
```

setter

```python
class Person:
    def __init__(self, name):
        self.full_name = name

    @property
    def name(self):
        print("Getting name...")
        return self.full_name

    @name.setter
    def name(self, value):
        print("Setting name...")
        self.full_name = value
```

deleter

```python
class Person:
    def __init__(self, name):
        self.full_name = name

    @property
    def name(self):
        print("Getting name...")
        return self.full_name

    @name.setter
    def name(self, value):
        print("Setting name...")
        self.full_name = value

    @name.deleter
    def name(self):
        print("Deleting name...")
        del self.full_name
```

```python
person = Person("Sagar Pun")
```

```python
person.name = "John Doe"
```

```python
person.name
```

```python
del person.name
```

```python
import uuid

class Person:
    def __init__(self, name):
        self.full_name = name

    @property
    def name(self):
        print("Getting name...")
        return self.full_name

    @property
    def id(self):
        print("Getting id...")
        return f"{self.full_name[-3:]}-{str(uuid.uuid4())[:8]}"
```

```python
obj = Person("Sagar Pun")
```

```python
obj = Person("Sagar Pun")
obj.id
```

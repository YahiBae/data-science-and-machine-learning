```python
class Person:
    count = 0
    def __init__(self, name: str,address: str = "kathmandu"):
        self.name = name
        self.address = address
        Person.count += 1
    def get_details(self):
        print(f"i am {self.name}, from {self.address}")

    @classmethod
    def get_count(cls):
        return cls.count

    @staticmethod
    def get_full_name(first, last):
        return f"{first} {last}"
```

```python
student1 = Person("sagar", "kathmandu")
```

```python
student1.get_full_name("sagar", "pun")
```

```python
class StudentResults:
    def __init__(self, name: str, mark1: int, mark2: int, mark3: int):
        self.name = name
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3

    def total(self):
        return self.mark1 + self.mark2 + self.mark3

    def average(self):
        return self.total() / 3

    def display(self):
        print(f"Student Name: {self.name}")
        print(f"Total Marks: {self.total()}")
        print(f"Average Marks: {self.average():.2f}")


student = StudentResults("Sagar", 85, 90, 78)
student.display()
```

```python
class PassCheck:
    def __init__(self, password: str):
        """class to check if the password is strong or weak

        parameters
        ----------
        password : str
            user input, password string to check
        """
        self.password = password

    def __str__(self):
        return f"{self.password[0]}***{self.password[-1]}"

    def is_strong(self):
        return (
            len(self.password) >= 8
            and any(char.isalpha() for char in self.password)
            and any(char.isdigit() for char in self.password)
        )


pass_check = PassCheck("hello23456")

print(pass_check)
print("Strong Password" if pass_check.is_strong() else "Weak Password")
```

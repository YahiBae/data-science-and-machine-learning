merge dict

```python
dict1 = {"name": "Sagar", "age": 23}
dict2 = {"city": "Kathmandu", "country": "Nepal"}

result = dict1 | dict2

print(result)
```

```python
dict1 = {"name": "Sagar", "age": 23}
dict2 = {"city": "Kathmandu", "country": "Nepal"}

result = {**dict1, **dict2}

print(result)
```

```python
dict1 = {"name": "Sagar", "age": 23}
dict2 = {"city": "Kathmandu", "country": "Nepal"}

dict1.update(dict2)

print(dict1)
```

```python
dict1 = {"name": "Sagar", "age": 23}
dict2 = {"city": "Kathmandu"}

result = dict1.copy()
result.update(dict2)

print(result)
```

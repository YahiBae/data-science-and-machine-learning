zip()

```python
name = ["John", "Alice", "Bob", "Eve"]
section = ["A", "B", "C", "D"]
sector = ["X", "Y", "Z",]
zip(name, section, sector)
```

```python
list(zip(name, section, sector))
```

```python
for a,b,c in zip(name, section, sector):
    print(f"{a} - {b} - {c}")
```

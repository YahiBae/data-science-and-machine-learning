method overloadiung

```python
class Example:
     def add(self, a, b,c=0):
            return a + b + c
```

```python
e =  Example()

print(e.add(2, 3))
print(e.add(2, 3, 4))
print(e.add("1", "2", "3"))
```

```python
class Example:
    def add(self, *args):
        return sum(args)
```

```python
ex = Example()
print(ex.add(1, 4, 6, 9, 8))
```

method overriding

```python
class Animal:
    def sound(self):
        print("making sound")
    
```

```python
class Dog(Animal):
    def sound(self):
        super().sound()
        print("barking")
```

```python
dog = Dog()

dog.sound()
```

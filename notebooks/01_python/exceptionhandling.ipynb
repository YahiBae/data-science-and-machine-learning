basic except handling in python

```python
try:
    number = int(input("Enter a number: "))
    print(6 / number)

except ValueError as e:
    print(e)
    print("invalid input")
    print(0)
```

catching multiple exceptions

```python
try:
    number = int(input("Enter a number: "))
    print(6 / number)

except ValueError as e:
    print(e)
    print("invalid input")
    print(0)

except ZeroDivisionError as e:
    print("cannot divide by zero")
    print(0)

except Exception as e:
    print("unknown error")
    print(0)

print("end of program")
```

```python
value = "abc"
try:
    print(int(value))

except (ValueError, TypeError) as e:
    print(f"{value} - cannot be converted to int", e)
```

the else block

```python
try:
    x = float(input("Enter a number: "))
    y = float(input("Enter another number: "))
    result = x / y

except ValueError as err:
    print("Invalid input. Please enter numeric values.")
    print(err)
    result = 1

else:
    print("final result", result)
```

the finally block

```python
try:
    x = float(input("Enter a number: "))
    y = float(input("Enter another number: "))
    result = x / y

except ValueError as err:
    print("Invalid input. Please enter numeric values.")
    print(err)
    result = 1

else:
    print("final result", result)

finally:
    print("end of program")
```

```python
lst = ["a", "b", "c", "d"]

def get_element(index):
    return lst[index]

# Try to read an element using a possibly invalid index.
try:
    print("Requested index 10:", get_element(10))
except IndexError:
    print("Index out of range. Returning last element.")
    if lst:
        print("Fallback value:", lst[-1])
    else:
        print("The list is empty, so no fallback value is available.")
```

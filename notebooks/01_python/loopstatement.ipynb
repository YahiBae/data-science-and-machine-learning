```python
adj = ["red","big","tasty"]
fruits = ["banana","apple","cherry"]
for fruit in fruits:
    for adjective in adj:
        print(f"{adjective} {fruit}")
```

```python
number = 8

for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

```python
# Python
# Input matrix:
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Expected Output: [4, 16, 36, 64]
# Your code here:
squared_evens = []
```

```python
for items in matrix:
    for num in items:
        if num % 2 == 0:
            squared_evens.append(num ** 2)
squared_evens
```

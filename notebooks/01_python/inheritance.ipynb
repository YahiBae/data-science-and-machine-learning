inheritance in python


1. single inheritance

```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f" Hello, I am {self.name}")
```

```python
class cat(Animal):
    def __init__(self, name, age, color):
        super().__init__(name, age)
        self.color = color

    def details(self):
        print(f"{self.color} cat : \n{self.name} \n{self.age}")
```

```python
cat = cat("tommy", 2, "black")
```

```python
cat.intro()
```

```python
cat.details()
```

multiple inheritance

```python
class Parent1:
    def func1(self):
        print("This is parent class 1")

class Parent2:
    def func1(self):
        print("This is parent class 2")

class Parent3:
    def func1(self):
        print("This is parent class 3")
```

```python
class Child(Parent1, Parent2, Parent3):
    def func4(self):
        print("This is child class")
```

```python
obj = Child()
```

```python
obj.func4()
```

```python
obj.func1()
```

multi-level inheritance

```python
class A:
    def method_a(self):
        print("This is a method from class A")

class B(A):
    def method_b(self):
        print("This is a method from class B")

class C(B):
    def method_c(self):
        print("This is a method from class C")
```

```python
b = C()
print(dir(b))
```

```python
c = C()
```

```python
c.method_c()
```

```python
c.method_c()
```

```python
c.method_b()
```

```python
class Person:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age


class Employee(Person):
    def __init__(self, name: str, age: int, employee_id: str):
        super().__init__(name, age)
        self.employee_id = employee_id


class Manager(Employee):
    def __init__(self, name: str, age: int, employee_id: str, department: str):
        super().__init__(name, age, employee_id)
        self.department = department

    def get_details(self):
        return (
            f"Name: {self.name}, Age: {self.age}, "
            f"Employee ID: {self.employee_id}, Department: {self.department}"
        )


manager = Manager("Sagar", 30, "EMP101", "Data Science")
print(manager.get_details())
```

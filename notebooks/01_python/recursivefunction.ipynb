recursive function

```python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
%%time
fib(5)
```

memorizing the recursive algorithm

```python
cache = {0 : 0, 1: 1}
def fib(n):
    if n in cache:
        return cache[n]
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        result = fib(n-1) + fib(n-2)
        cache[n] = result
        return result
```

```python
%%time
fib(100)
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

```python
factorial(5)
```

```python
cache = [1,1]
def factorial(n):
    if n < len(cache):
        return cache[n]
    else:
        result = n * factorial(n-1)
        cache.append(result)
        return result
```

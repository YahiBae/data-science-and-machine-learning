magic methods in py

_init_
_repr_
_str_
_len_
_call_

```python
class Person:
    """
    hello
    """
    def __init__(self,name,age):
        self.name = name
        self.age = age 

    def __str__(self):
        return f"{self.name} / {self.age}"

    def __repr__(self):
        return f"Person({self.name}, {self.age})"

    def __len__(self):
        return self.age

    def __call__(self, gender = "male"):
        return f"{self.name} is a {gender}."
```

```python
person = Person("Sagar", 25)    
```

```python
print(person)
```

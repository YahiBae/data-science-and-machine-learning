```python
test_numbers = [
    -10, -1, 0, 1, 2, 3, 4, 5, 6, 7, 9, 11, 13, 15, 17, 19, 21, 25,
    27, 47, 49, 79, 97, 100, 523, 527, 1000, 7919, 10000, 104729, 1000000
]


def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


prime_numbers = [number for number in test_numbers if is_prime(number)]
print(prime_numbers)
```

```python
def print_all(*args):
    print("positional args")
    print(args)
```

```python
print_all(1, 2, 3, 4, 5,"a", True)
```

```python
def sum_all(*nums):
    return sum(nums)
```

```python

print(sum_all(1, 2, 3, 4, 5,6.8))
```

```python
def print_kwargs(**kwargs):
    print("keyword args")
    print(kwargs)
```

```python
print_kwargs(num1=1,num2=2)
```

```python
def calculate_area(shape:str, **kwargs):
    if shape == "square":
        area = kwargs.get("side") ** 2
    elif shape == "rectangle":
        if "length" in kwargs.keys() and "breadth" in kwargs.keys():
            area = kwargs.get("length", 0) * kwargs.get("breadth", 0)
    else:
        print(f"{shape} is not supported")
        area = 0
    return area
```

```python
calculate_area("rectangle", length=5, breadth=3)
```

```python
def print_args(shape, *args, **kwargs):
    print(shape)
    print(args)
    print(kwargs)
```

```python
print_args("circle", 1, 2, 3, radius=5, color="red")
```

```python

import statistics


def std_deviation(numbers):
    return statistics.pstdev(numbers)

numbers = [10, 12, 15, 18, 20]

print("Standard Deviation:", std_deviation(numbers))
```

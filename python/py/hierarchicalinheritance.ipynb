hierarchical inheritance


```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f" Hello, I am {self.name}")
```

```python
class Cat(Animal):
    def speak(self):
        print("Meow!")

class Dog(Animal):
    def speak(self):
        print("Woof!")

class lion(Animal):
    def speak(self):
        print("Roar!")
```

```python
cat = Cat("Whiskers", 3)
```

```python
print(cat.intro())
```

```python
print(cat.speak())
```

```python
dog = Dog("Buddy", 5)
```

```python
dog.speak()
```

hybrid inheritance 

```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f" Hello, I am {self.name}")
```

```python
class Mammal(Animal):
    def __init__(self, name, age, fur =False):
        super().__init__(name, age)
        self.fur = fur

    def birth(self):
        print(f"{self.name} is giving birth.")

class Birds(Animal):
    def fly(self):
        return f"{self.name} is flying."
```

```python
class Bat(Mammal, Birds):
    pass
```

```python
b = Bat("Bruce", 2, True)
```

```python
b.intro()
```

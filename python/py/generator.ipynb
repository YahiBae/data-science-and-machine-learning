python generators

```python
def number_generator(start, end):
    for number in range(start, end + 1):
        yield number
```

```python
gen = number_generator(1, 10)
```

```python
print(gen)
```

```python
next(gen)
```

```python
for number in gen:
    print(number)
```

```python
my_gen = number_generator(1, 10)
print(list(my_gen))
```

```python
def is_prime( n):
        if n < 2:
            return False
        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False
        return True
```

```python
def Prime_generator(limit = 10):
    count, start = 0, 2

    while count < limit:
        if is_prime(start):
            yield start
            count += 1
        start += 1
```

```python
prime_gen_obj = Prime_generator(10)
```

```python
list(prime_gen_obj)
```

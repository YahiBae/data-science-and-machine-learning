```python
arr1 = np.random.randint(1, 100, size =  (3, 3))
arr2 = np.random.randint(1, 100, size =  (3, 3))    
```

```python
arr1, arr2
```

```python
np.concatenate((arr1, arr2))
```

```python
np.concatenate((arr1, arr2), axis = 0) # vstack
```

```python
np.vstack((arr1, arr2))
```

```python
np.concatenate((arr1, arr2), axis = 1) # hstack
```

```python
np.hstack((arr1, arr2))
```
